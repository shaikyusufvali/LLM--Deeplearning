# Similarity Search

```python
import pandas as pd
import numpy as np
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity

df = pd.read_csv("../data/llm_cleaned_dataset.csv")

model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
def similarity_search(query, top_k=5):

    query_embedding = model.encode([query])

    scores = cosine_similarity(
        query_embedding,
        document_embeddings
    )[0]

    top_indices = np.argsort(scores)[::-1][:top_k]

    results = []

    for index in top_indices:
        results.append({
            "doc_id": df.iloc[index]["doc_id"],
            "topic": df.iloc[index]["topic"],
            "similarity": round(float(scores[index]), 4),
            "text": df.iloc[index]["clean_text"]
        })

    return pd.DataFrame(results)
```

```python
query = "What is machine learning?"

results = similarity_search(query, top_k=5)

results
```

# chunking

```python
def create_chunks(text, chunk_size=40):

    words = text.split()

    chunks = []

    for i in range(0, len(words), chunk_size):

        chunk = words[i:i + chunk_size]

        chunks.append(" ".join(chunk))

    return chunks
```

```python
text = df["clean_text"].iloc[0]

chunks = create_chunks(text, chunk_size=40)

for i, chunk in enumerate(chunks):
    print(f"CHUNK {i + 1}:")
    print(chunk)
    print("-" * 50)
```

```python
all_chunks = []

for _, row in df.iterrows():

    chunks = create_chunks(
        row["clean_text"],
        chunk_size=40
    )

    for i, chunk in enumerate(chunks):

        all_chunks.append({
            "doc_id": row["doc_id"],
            "topic": row["topic"],
            "chunk_id": i,
            "chunk_text": chunk
        })

chunks_df = pd.DataFrame(all_chunks)

chunks_df.head(10)
```

# Chunk Overlap

```python
# Create chunking function with overlap
def create_chunks_with_overlap(text, chunk_size=40, overlap=10):

    if overlap >= chunk_size:
        raise ValueError("Overlap must be smaller than chunk_size")

    words = text.split()
    chunks = []

    step = chunk_size - overlap

    for i in range(0, len(words), step):

        chunk = words[i:i + chunk_size]

        if not chunk:
            break

        chunks.append(" ".join(chunk))

    return chunks
```

```python
text = df["clean_text"].iloc[0]

chunks = create_chunks_with_overlap(
    text,
    chunk_size=40,
    overlap=10
)

for i, chunk in enumerate(chunks):
    print(f"\nCHUNK {i + 1}")
    print(chunk)
    print("Word Count:", len(chunk.split()))
    print("-" * 60)
```

```python
# Compare without overlap vs with overlap
text = df["clean_text"].iloc[0]

normal_chunks = create_chunks_with_overlap(
    text,
    chunk_size=40,
    overlap=0
)

overlap_chunks = create_chunks_with_overlap(
    text,
    chunk_size=40,
    overlap=10
)

print("Without Overlap:", len(normal_chunks))
print("With Overlap:", len(overlap_chunks))
```

```python
# Apply to all 500 documents
all_chunks = []

for _, row in df.iterrows():

    chunks = create_chunks_with_overlap(
        row["clean_text"],
        chunk_size=40,
        overlap=10
    )

    for i, chunk in enumerate(chunks):

        all_chunks.append({
            "doc_id": row["doc_id"],
            "topic": row["topic"],
            "chunk_id": i,
            "chunk_text": chunk
        })

chunks_df = pd.DataFrame(all_chunks)

print("Original Documents:", len(df))
print("Total Chunks:", len(chunks_df))

chunks_df.head()
```

# TF-IDF

```python
tfidf = TfidfVectorizer()

tfidf_matrix = tfidf.fit_transform(
    chunks_df["chunk_text"].fillna("")
)

def keyword_search(query, top_k=5):

    query_vector = tfidf.transform([query])

    scores = cosine_similarity(
        query_vector,
        tfidf_matrix
    )[0]

    top_indices = np.argsort(scores)[::-1][:top_k]

    results = chunks_df.iloc[top_indices].copy()
    results["keyword_score"] = scores[top_indices]

    return results
```

```python
def semantic_search(query, top_k=5):

    query_embedding = model.encode([query]).tolist()

    results = collection.query(
        query_embeddings=query_embedding,
        n_results=top_k
    )

    documents = results["documents"][0]
    metadatas = results["metadatas"][0]
    distances = results["distances"][0]

    return documents, metadatas, distances
```

```python
def hybrid_search(query, top_k=5):

    # Keyword Search
    keyword_results = keyword_search(query, top_k=top_k)

    # Semantic Search
    semantic_documents, semantic_metadata, semantic_distances = (
        semantic_search(query, top_k=top_k)
    )

    # Combine results
    combined = {}

    for _, row in keyword_results.iterrows():

        doc_id = str(row["doc_id"])
        chunk_id = int(row["chunk_id"])

        key = (doc_id, chunk_id)

        combined[key] = {
            "doc_id": doc_id,
            "topic": row["topic"],
            "chunk_id": chunk_id,
            "chunk_text": row["chunk_text"],
            "keyword_score": float(row["keyword_score"]),
            "semantic_score": 0.0
        }

    for document, metadata, distance in zip(
        semantic_documents,
        semantic_metadata,
        semantic_distances
    ):

        key = (
            str(metadata["doc_id"]),
            int(metadata["chunk_id"])
        )

        if key not in combined:

            combined[key] = {
                "doc_id": key[0],
                "topic": metadata["topic"],
                "chunk_id": key[1],
                "chunk_text": document,
                "keyword_score": 0.0,
                "semantic_score": 0.0
            }

        # Convert cosine distance into similarity
        combined[key]["semantic_score"] = max(0.0, 1.0 - float(distance))

    results_df = pd.DataFrame(combined.values())

    if results_df.empty:
        return results_df

    # Normalize keyword scores
    max_score = results_df["keyword_score"].max()

    if max_score > 0:
        results_df["keyword_score"] = (
            results_df["keyword_score"] / max_score
        )

    # Combine both scores
    results_df["hybrid_score"] = (
        0.5 * results_df["keyword_score"]
        + 0.5 * results_df["semantic_score"]
    )

    return results_df.sort_values(
        by="hybrid_score",
        ascending=False
    ).head(top_k)
```

```python
query = "How do machines learn from data?"

results = hybrid_search(query, top_k=5)

results[[
    "topic",
    "keyword_score",
    "semantic_score",
    "hybrid_score",
    "chunk_text"
]]
```

```python
# Few-Shot: Multiple examples provide chestham

few_shot_prompt = f"""
Task: Classify the text topic.

Example 1:
Text: Robots use sensors and motors.
Topic: Robotics

Example 2:
Text: Neural networks contain artificial neurons.
Topic: Deep Learning

Example 3:
Text: Computers process human language.
Topic: NLP

Now classify:
Text: {sample_text}

Return only the topic name.
"""

print(few_shot_prompt)
```

# System Prompt vs User Prompt

```python
# System Prompt vs User Prompt
# System Prompt: LLM role and behaviour

system_prompt = """
You are an expert AI and Machine Learning teacher.
Give accurate and beginner-friendly explanations.
"""

# User Prompt: User task or question

user_prompt = """
Explain Machine Learning with one real-world example.
"""

print("SYSTEM PROMPT:\n", system_prompt)
print("\nUSER PROMPT:\n", user_prompt)
```

# Prompt Templates

```python
# Prompt Template: Reusable prompt with placeholders

def create_prompt(topic, level):
    prompt = f"""
    You are an AI teacher.

    Explain {topic} for a {level} learner.

    Include:
    1. Definition
    2. Working process
    3. Real-world example
    """
    return prompt


# Reuse the same template with different inputs
prompt_1 = create_prompt("Machine Learning", "beginner")
prompt_2 = create_prompt("Deep Learning", "intermediate")

print(prompt_1)
print(prompt_2)
```

```python
from openai import OpenAI

client = OpenAI(
    api_key=groq_key,
    base_url="https://api.groq.com/openai/v1"
)

response = client.chat.completions.create(
    model="openai/gpt-oss-20b",
    messages=[
        {
            "role": "user",
            "content": "Explain Large Language Models in 3 simple sentences."
        }
    ]
)

print(response.choices[0].message.content)
```

```python
def generate_llm_response(prompt, model="openai/gpt-oss-20b"):

    response = client.chat.completions.create(
        model=model,
        messages=[
            {
                "role": "user",
                "content": prompt
            }
        ],
        temperature=0.3,
        max_tokens=300
    )

    return response.choices[0].message.content
```

```python
prompt = "Explain RAG in simple English with one example."

answer = generate_llm_response(prompt)

print(answer)
```
