# Similarity Search

```python
import pandas as pd
import numpy as np
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity

df = pd.read_csv("../data/llm_cleaned_dataset.csv")

model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
def similarity_search(query, top_k=5):

    query_embedding = model.encode([query])

    scores = cosine_similarity(
        query_embedding,
        document_embeddings
    )[0]

    top_indices = np.argsort(scores)[::-1][:top_k]

    results = []

    for index in top_indices:
        results.append({
            "doc_id": df.iloc[index]["doc_id"],
            "topic": df.iloc[index]["topic"],
            "similarity": round(float(scores[index]), 4),
            "text": df.iloc[index]["clean_text"]
        })

    return pd.DataFrame(results)
```

```python
query = "What is machine learning?"

results = similarity_search(query, top_k=5)

results
```

# chunking

```python
def create_chunks(text, chunk_size=40):

    words = text.split()

    chunks = []

    for i in range(0, len(words), chunk_size):

        chunk = words[i:i + chunk_size]

        chunks.append(" ".join(chunk))

    return chunks
```

```python
text = df["clean_text"].iloc[0]

chunks = create_chunks(text, chunk_size=40)

for i, chunk in enumerate(chunks):
    print(f"CHUNK {i + 1}:")
    print(chunk)
    print("-" * 50)
```

```python
all_chunks = []

for _, row in df.iterrows():

    chunks = create_chunks(
        row["clean_text"],
        chunk_size=40
    )

    for i, chunk in enumerate(chunks):

        all_chunks.append({
            "doc_id": row["doc_id"],
            "topic": row["topic"],
            "chunk_id": i,
            "chunk_text": chunk
        })

chunks_df = pd.DataFrame(all_chunks)

chunks_df.head(10)
```

# Chunk Overlap

```python
# Create chunking function with overlap
def create_chunks_with_overlap(text, chunk_size=40, overlap=10):

    if overlap >= chunk_size:
        raise ValueError("Overlap must be smaller than chunk_size")

    words = text.split()
    chunks = []

    step = chunk_size - overlap

    for i in range(0, len(words), step):

        chunk = words[i:i + chunk_size]

        if not chunk:
            break

        chunks.append(" ".join(chunk))

    return chunks
```

```python
text = df["clean_text"].iloc[0]

chunks = create_chunks_with_overlap(
    text,
    chunk_size=40,
    overlap=10
)

for i, chunk in enumerate(chunks):
    print(f"\nCHUNK {i + 1}")
    print(chunk)
    print("Word Count:", len(chunk.split()))
    print("-" * 60)
```

```python
# Compare without overlap vs with overlap
text = df["clean_text"].iloc[0]

normal_chunks = create_chunks_with_overlap(
    text,
    chunk_size=40,
    overlap=0
)

overlap_chunks = create_chunks_with_overlap(
    text,
    chunk_size=40,
    overlap=10
)

print("Without Overlap:", len(normal_chunks))
print("With Overlap:", len(overlap_chunks))
```

```python
# Apply to all 500 documents
all_chunks = []

for _, row in df.iterrows():

    chunks = create_chunks_with_overlap(
        row["clean_text"],
        chunk_size=40,
        overlap=10
    )

    for i, chunk in enumerate(chunks):

        all_chunks.append({
            "doc_id": row["doc_id"],
            "topic": row["topic"],
            "chunk_id": i,
            "chunk_text": chunk
        })

chunks_df = pd.DataFrame(all_chunks)

print("Original Documents:", len(df))
print("Total Chunks:", len(chunks_df))

chunks_df.head()
```

# TF-IDF

```python
tfidf = TfidfVectorizer()

tfidf_matrix = tfidf.fit_transform(
    chunks_df["chunk_text"].fillna("")
)

def keyword_search(query, top_k=5):

    query_vector = tfidf.transform([query])

    scores = cosine_similarity(
        query_vector,
        tfidf_matrix
    )[0]

    top_indices = np.argsort(scores)[::-1][:top_k]

    results = chunks_df.iloc[top_indices].copy()
    results["keyword_score"] = scores[top_indices]

    return results
```

```python
def semantic_search(query, top_k=5):

    query_embedding = model.encode([query]).tolist()

    results = collection.query(
        query_embeddings=query_embedding,
        n_results=top_k
    )

    documents = results["documents"][0]
    metadatas = results["metadatas"][0]
    distances = results["distances"][0]

    return documents, metadatas, distances
```

```python
def hybrid_search(query, top_k=5):

    # Keyword Search
    keyword_results = keyword_search(query, top_k=top_k)

    # Semantic Search
    semantic_documents, semantic_metadata, semantic_distances = (
        semantic_search(query, top_k=top_k)
    )

    # Combine results
    combined = {}

    for _, row in keyword_results.iterrows():

        doc_id = str(row["doc_id"])
        chunk_id = int(row["chunk_id"])

        key = (doc_id, chunk_id)

        combined[key] = {
            "doc_id": doc_id,
            "topic": row["topic"],
            "chunk_id": chunk_id,
            "chunk_text": row["chunk_text"],
            "keyword_score": float(row["keyword_score"]),
            "semantic_score": 0.0
        }

    for document, metadata, distance in zip(
        semantic_documents,
        semantic_metadata,
        semantic_distances
    ):

        key = (
            str(metadata["doc_id"]),
            int(metadata["chunk_id"])
        )

        if key not in combined:

            combined[key] = {
                "doc_id": key[0],
                "topic": metadata["topic"],
                "chunk_id": key[1],
                "chunk_text": document,
                "keyword_score": 0.0,
                "semantic_score": 0.0
            }

        # Convert cosine distance into similarity
        combined[key]["semantic_score"] = max(0.0, 1.0 - float(distance))

    results_df = pd.DataFrame(combined.values())

    if results_df.empty:
        return results_df

    # Normalize keyword scores
    max_score = results_df["keyword_score"].max()

    if max_score > 0:
        results_df["keyword_score"] = (
            results_df["keyword_score"] / max_score
        )

    # Combine both scores
    results_df["hybrid_score"] = (
        0.5 * results_df["keyword_score"]
        + 0.5 * results_df["semantic_score"]
    )

    return results_df.sort_values(
        by="hybrid_score",
        ascending=False
    ).head(top_k)
```

```python
query = "How do machines learn from data?"

results = hybrid_search(query, top_k=5)

results[[
    "topic",
    "keyword_score",
    "semantic_score",
    "hybrid_score",
    "chunk_text"
]]
```

```python
def create_rag_prompt(query, context):

    prompt = f"""
    You are an AI assistant using Retrieval Augmented Generation.

    Answer the question using ONLY the provided context.

    If the answer is not available in the context,
    clearly say that the information is not available.

    CONTEXT:
    {context}

    QUESTION:
    {query}

    INSTRUCTIONS:
    1. Give a clear and accurate answer.
    2. Do not invent information.
    3. Keep the answer easy to understand.
    4. Mention relevant document sources.

    ANSWER:
    """

    return prompt
```

# Complete RAG Pipeline

```python
def rag_pipeline(query, retrieved_results, top_k=5):

    # Step 1: Reranking
    reranked_results = rerank_results(
        query,
        retrieved_results,
        top_k
    )

    # Step 2: Context building
    context = build_context(
        reranked_results,
        top_k
    )

    # Step 3: Prompt creation
    prompt = create_rag_prompt(
        query,
        context
    )

    # Step 4: LLM generation
    answer = safe_llm_response(prompt)

    # Step 5: Source extraction
    sources = (
        reranked_results[
            ["doc_id", "topic", "chunk_id", "rerank_score"]
        ].to_dict(orient="records")
        if not reranked_results.empty
        else []
    )

    return {
        "question": query,
        "answer": answer,
        "sources": sources
    }
```

```python
import numpy as np

def precision_at_k(retrieved, relevant, k=5):

    retrieved_k = retrieved[:k]

    if not retrieved_k:
        return 0.0

    relevant_count = sum(
        doc_id in relevant
        for doc_id in retrieved_k
    )

    return relevant_count / k


def recall_at_k(retrieved, relevant, k=5):

    if not relevant:
        return 0.0

    retrieved_k = retrieved[:k]

    relevant_count = sum(
        doc_id in relevant
        for doc_id in retrieved_k
    )

    return relevant_count / len(relevant)


def reciprocal_rank(retrieved, relevant):

    for rank, doc_id in enumerate(retrieved, start=1):

        if doc_id in relevant:
            return 1 / rank

    return 0.0
```

```python
def ndcg_at_k(retrieved, relevant, k=5):

    relevance = [
        1 if doc_id in relevant else 0
        for doc_id in retrieved[:k]
    ]

    dcg = sum(
        rel / np.log2(index + 2)
        for index, rel in enumerate(relevance)
    )

    ideal = sorted(relevance, reverse=True)

    idcg = sum(
        rel / np.log2(index + 2)
        for index, rel in enumerate(ideal)
    )

    return dcg / idcg if idcg > 0 else 0.0
```

# Complete RAG Implementation

```python
import os
import time
import numpy as np
import pandas as pd
import chromadb

from pathlib import Path
from dotenv import load_dotenv
from openai import (
    OpenAI,
    RateLimitError,
    APIConnectionError,
    APITimeoutError,
    APIStatusError
)

from sentence_transformers import (
    SentenceTransformer,
    CrossEncoder
)

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Load environment variables
env_path = Path("../.env").resolve()
load_dotenv(env_path, override=True)

# Groq API connection
groq_key = os.getenv("GROQ_API_KEY")

if not groq_key:
    raise ValueError("GROQ_API_KEY missing in .env")

client = OpenAI(
    api_key=groq_key,
    base_url="https://api.groq.com/openai/v1",
    timeout=30.0,
    max_retries=0
)

# Load chunks
chunks_df = pd.read_csv("../data/llm_chunks_overlap.csv")

# ChromaDB connection
chroma_client = chromadb.PersistentClient(
    path="../data/chroma_db"
)

collection = chroma_client.get_collection(
    name="llm_practice_chunks_v1"
)

# Embedding model
embedding_model = SentenceTransformer(
    "all-MiniLM-L6-v2"
)

# Cross-Encoder
cross_encoder = CrossEncoder(
    "cross-encoder/ms-marco-MiniLM-L-6-v2"
)

print("All components initialized successfully!")
print("Total chunks:", len(chunks_df))
print("ChromaDB chunks:", collection.count())
```

```python
def generate_llm_response(
    prompt,
    temperature=0.2,
    max_tokens=1000
):

    response = client.chat.completions.create(
        model="openai/gpt-oss-20b",
        messages=[
            {
                "role": "system",
                "content": "You are a helpful AI assistant."
            },
            {
                "role": "user",
                "content": prompt
            }
        ],
        temperature=temperature,
        max_tokens=max_tokens
    )

    return response.choices[0].message.content
```

```python
def safe_llm_response(prompt, retries=3):

    for attempt in range(retries):

        try:

            return generate_llm_response(prompt)

        except RateLimitError:

            wait_time = 2 ** attempt

            print(f"Rate limit. Retrying in {wait_time} seconds.")

            time.sleep(wait_time)

        except APITimeoutError:

            print("API timeout.")

            time.sleep(2)

        except APIConnectionError:

            print("Connection error.")
            return None

        except APIStatusError as error:

            print("API Status:", error.status_code)
            return None

        except Exception as error:

            print("Unexpected error:", str(error))
            return None

    print("Maximum retry attempts reached.")

    return None
```

```python
tfidf_vectorizer = TfidfVectorizer(
    max_features=10000,
    stop_words="english"
)

tfidf_matrix = tfidf_vectorizer.fit_transform(
    chunks_df["chunk_text"].fillna("")
)

print("TF-IDF index created!")
```

```python
def hybrid_search(query, top_k=10):

    # Semantic retrieval
    query_embedding = embedding_model.encode(
        query
    ).tolist()

    semantic_results = collection.query(
        query_embeddings=[query_embedding],
        n_results=top_k
    )

    semantic_docs = []

    for rank, (doc, metadata) in enumerate(
        zip(
            semantic_results["documents"][0],
            semantic_results["metadatas"][0]
        ),
        start=1
    ):

        semantic_docs.append({
            "doc_id": metadata["doc_id"],
            "topic": metadata["topic"],
            "chunk_id": metadata["chunk_id"],
            "chunk_text": doc,
            "semantic_rank": rank
        })

    # Keyword retrieval
    query_vector = tfidf_vectorizer.transform([query])

    scores = cosine_similarity(
        query_vector,
        tfidf_matrix
    ).flatten()

    keyword_indices = np.argsort(scores)[::-1][:top_k]

    keyword_docs = []

    for rank, index in enumerate(keyword_indices, start=1):

        row = chunks_df.iloc[index]

        keyword_docs.append({
            "doc_id": row["doc_id"],
            "topic": row["topic"],
            "chunk_id": row["chunk_id"],
            "chunk_text": row["chunk_text"],
            "keyword_rank": rank
        })

    # Reciprocal Rank Fusion
    combined = {}

    for item in semantic_docs:

        key = (item["doc_id"], int(item["chunk_id"]))

        combined[key] = {
            **item,
            "rrf_score": 1 / (60 + item["semantic_rank"])
        }

    for item in keyword_docs:

        key = (item["doc_id"], int(item["chunk_id"]))

        if key in combined:

            combined[key]["rrf_score"] += (
                1 / (60 + item["keyword_rank"])
            )

        else:

            combined[key] = {
                **item,
                "rrf_score": 1 / (60 + item["keyword_rank"])
            }

    results = pd.DataFrame(list(combined.values()))

    results = results.sort_values(
        "rrf_score",
        ascending=False
    )

    return results.head(top_k).reset_index(drop=True)
```

```python
def rerank_results(query, retrieved_results, top_k=5):

    if retrieved_results is None or retrieved_results.empty:
        return pd.DataFrame()

    rerank_df = retrieved_results.copy()

    pairs = [
        (query, text)
        for text in rerank_df["chunk_text"].tolist()
    ]

    scores = cross_encoder.predict(pairs)

    rerank_df["rerank_score"] = scores

    rerank_df = rerank_df.sort_values(
        by="rerank_score",
        ascending=False
    )

    return rerank_df.head(top_k).reset_index(drop=True)
```

```python
def build_context(retrieved_results):

    if retrieved_results.empty:
        return "No relevant information found."

    context_parts = []

    for _, row in retrieved_results.iterrows():

        context = f"""
[Source: {row['doc_id']}]
[Topic: {row['topic']}]
[Chunk: {row['chunk_id']}]

{row['chunk_text']}
"""

        context_parts.append(context)

    return "\n\n".join(context_parts)
```

```python
def rag_pipeline(query, top_k=5):

    # Step 1: Hybrid Search
    retrieved_results = hybrid_search(
        query,
        top_k=10
    )

    if retrieved_results.empty:

        return {
            "question": query,
            "answer": "No relevant information found.",
            "sources": []
        }

    # Step 2: Reranking
    reranked_results = rerank_results(
        query,
        retrieved_results,
        top_k=top_k
    )

    # Step 3: Context
    context = build_context(
        reranked_results
    )

    # Step 4: Prompt
    prompt = create_rag_prompt(
        query,
        context
    )

    # Step 5: LLM
    answer = safe_llm_response(prompt)

    # Step 6: Source details
    sources = []

    for _, row in reranked_results.iterrows():

        sources.append({
            "doc_id": row["doc_id"],
            "topic": row["topic"],
            "chunk_id": int(row["chunk_id"]),
            "rerank_score": float(row["rerank_score"])
        })

    return {
        "question": query,
        "answer": answer,
        "sources": sources
    }
```

```python
def precision_at_k(retrieved, relevant, k=5):

    retrieved_k = retrieved[:k]

    if not retrieved_k:
        return 0.0

    hits = sum(
        doc in relevant
        for doc in retrieved_k
    )

    return hits / k


def recall_at_k(retrieved, relevant, k=5):

    if not relevant:
        return 0.0

    hits = sum(
        doc in relevant
        for doc in retrieved[:k]
    )

    return hits / len(relevant)


def reciprocal_rank(retrieved, relevant):

    for rank, doc in enumerate(retrieved, start=1):

        if doc in relevant:
            return 1 / rank

    return 0.0


def ndcg_at_k(retrieved, relevant, k=5):

    relevance = [
        1 if doc in relevant else 0
        for doc in retrieved[:k]
    ]

    dcg = sum(
        rel / np.log2(index + 2)
        for index, rel in enumerate(relevance)
    )

    ideal = sorted(relevance, reverse=True)

    idcg = sum(
        rel / np.log2(index + 2)
        for index, rel in enumerate(ideal)
    )

    return dcg / idcg if idcg > 0 else 0.0
```

```python
def evaluate_rag(evaluation_data, k=5):

    evaluation_results = []

    for item in evaluation_data:

        relevant = set(item["relevant_docs"])

        if not relevant:
            print(
                "Skipping: Ground truth missing for",
                item["question"]
            )
            continue

        retrieved_df = hybrid_search(
            item["question"],
            top_k=10
        )

        reranked_df = rerank_results(
            item["question"],
            retrieved_df,
            top_k=k
        )

        retrieved_ids = (
            reranked_df["doc_id"].tolist()
        )

        evaluation_results.append({

            "Question": item["question"],

            "Precision@K": precision_at_k(
                retrieved_ids, relevant, k
            ),

            "Recall@K": recall_at_k(
                retrieved_ids, relevant, k
            ),

            "MRR": reciprocal_rank(
                retrieved_ids, relevant
            ),

            "NDCG@K": ndcg_at_k(
                retrieved_ids, relevant, k
            )

        })

    evaluation_df = pd.DataFrame(
        evaluation_results
    )

    return evaluation_df
```
